# Day 2: Rotation Matrices & Coordinate Transformations
## Why robots need to transform sensor data

A robot’s sensors measure the world **relative to the robot itself** (local frame). To navigate, the robot must convert those measurements into a **world coordinate frame**.

In this notebook you will:
- Implement 2D rotation matrices.
- Learn the difference between **local** and **world** frames.
- Simulate two robots chasing a target:
  - **Wrong robot**: uses sensor measurements directly as world coordinates → fails.
  - **Correct robot**: rotates local measurements into world coordinates → succeeds.
- See why the right‑hand rule and unit circle define the rotation matrix.

This builds intuition for sensor fusion, extrinsic calibration, and VIO.

## 1. Recap of Day 1: vector functions
We reuse the same pure‑Python implementations (no NumPy).

````python
import math
import time
from IPython.display import clear_output

# ----- Day 1 functions (without numpy) -----
def dot_product(v1, v2):
    return sum(a*b for a,b in zip(v1,v2))

def norm(v):
    return math.sqrt(sum(x*x for x in v))

def unit_vector(v):
    n = norm(v)
    if n == 0: raise ValueError("Zero vector")
    return [x/n for x in v]

def matrix_multiply(A, B):
    rows_A, cols_A = len(A), len(A[0])
    rows_B, cols_B = len(B), len(B[0])
    if cols_A != rows_B: raise ValueError("Incompatible")
    result = [[0 for _ in range(cols_B)] for _ in range(rows_A)]
    for i in range(rows_A):
        for j in range(cols_B):
            total = 0
            for k in range(cols_A):
                total += A[i][k] * B[k][j]
            result[i][j] = total
    return result

def matrix_vector_multiply(A, v):
    rows, cols = len(A), len(A[0])
    if cols != len(v): raise ValueError("Matrix columns != vector length")
    return [sum(A[i][j] * v[j] for j in range(cols)) for i in range(rows)]
````

## 2. Derivation of the 2D Rotation Matrix

### What does a rotation matrix do?
A 2D rotation matrix `R(θ)` rotates any vector by angle θ **counter‑clockwise** around the origin.

### How to build the matrix: basis vectors
Any vector `(x, y)` can be expressed as `x·e₁ + y·e₂`, where:
- `e₁ = (1, 0)` is the unit vector along the X‑axis.
- `e₂ = (0, 1)` is the unit vector along the Y‑axis.

If we know where `e₁` and `e₂` go after rotation, we can rotate any vector.

### Step 1 – Where does `e₁ = (1, 0)` go?
Rotating (1,0) by angle θ places its tip on the unit circle at angle θ. By definition of cosine and sine:
```
R(θ)·(1,0) = (cosθ, sinθ)
```
This becomes the **first column** of the matrix.

### Step 2 – Where does `e₂ = (0, 1)` go?
The original Y‑axis points at 90°. Rotating it by θ gives an angle of 90°+θ.
Using trigonometric identities:
- `cos(90°+θ) = -sinθ`
- `sin(90°+θ) = cosθ`
So:
```
R(θ)·(0,1) = (-sinθ, cosθ)
```
This becomes the **second column**.

### Step 3 – Assemble the matrix
```
R(θ) = [ [cosθ, -sinθ],
         [sinθ,  cosθ] ]
```

### What do rows and columns represent?
- **Columns** = where the original X and Y axes land (transformed basis vectors).
- **Rows** = output coordinates (first row = X coordinate, second row = Y coordinate).

**Intuition:** Multiply the matrix by a vector `(x,y)`. The result is `x` times the first column plus `y` times the second column – that is exactly the linear combination of the rotated axes.

### Why counter‑clockwise? Convention.
Mathematics and robotics use positive θ for counter‑clockwise rotation (right‑hand rule in 3D). For clockwise rotation, use `R(-θ)`.

### The formula works for any θ (in degrees or radians).
Just compute cos and sin.

````python
def rotation_matrix_2d(angle_deg):
    """
    Returns a 2x2 rotation matrix for a given angle in degrees.
    Converts to radians because math.cos/sin use radians.
    """
    theta = math.radians(angle_deg)
    cos_t = math.cos(theta)
    sin_t = math.sin(theta)
    return [[cos_t, -sin_t],
            [sin_t,  cos_t]]

# Test: rotate (1,0) by 90°
R90 = rotation_matrix_2d(90)
point = [1,0]
rotated = matrix_vector_multiply(R90, point)
print(f"Rotating (1,0) by 90° → {rotated}")  # should be (0,1)
````

## 3. Transforming a point from local to world frame

Assume:
- Robot at `(rx, ry)` in world coordinates.
- Robot heading = θ (0 = east, CCW).
- Sensor gives local target vector `(dx_local, dy_local)` – e.g., (3,0) means 3 units straight ahead.

**To get world coordinates of that target:**
1. Rotate the local vector by θ (aligns it with world axes).
2. Add robot’s position.

Formula: `target_world = R(θ) · [dx_local, dy_local] + [rx, ry]`

**Why rotation?** Because the robot’s “forward” direction in world coordinates changes with its heading.

````python
def local_to_world(local_point, robot_pos, robot_heading_deg):
    """
    Convert a point given in robot-local coordinates to world coordinates.
    - local_point: [dx, dy] in robot frame
    - robot_pos: [rx, ry] world position of robot
    - robot_heading_deg: robot's orientation (0 = east, CCW)
    """
    R = rotation_matrix_2d(robot_heading_deg)          # step 1: get rotation matrix
    rotated = matrix_vector_multiply(R, local_point)   # step 2: rotate local vector
    world_point = [rotated[0] + robot_pos[0],          # step 3: translate
                   rotated[1] + robot_pos[1]]
    return world_point
````

## 4. Why the wrong robot fails – conceptual explanation

The **wrong robot** ignores its own heading. It takes the local measurement `[dx, dy]` (e.g., [4,0]) and treats it as **world coordinates** directly.  
So it will always try to move toward a fixed point in the world (e.g., (4,0)), regardless of where the robot actually is or how it is rotated.

The **correct robot** uses the transformation: it rotates the local vector and adds its position. Hence the target appears in the correct world location relative to the robot.

**Consequence:** The wrong robot may still move (it always has a target), but that target is **not** the real target. It will wander off and never reach the true target unless the robot happens to be exactly oriented east and positioned at the origin.

## 5. Simulation: both robots move toward the same target

We give both robots the **same local sensor measurement**.  
- **Wrong robot** → `target_world_wrong = local_measurement` (ignores heading).
- **Correct robot** → `target_world_correct = local_to_world(...)`.

Then each robot moves one step toward its target estimate.

````python
# Helper: draw robots, target on a grid
def draw_map(wrong_pos, correct_pos, target_pos, width=12, height=8):
    grid = [['.' for _ in range(width)] for _ in range(height)]
    wx, wy = int(round(wrong_pos[0])), int(round(wrong_pos[1]))
    cx, cy = int(round(correct_pos[0])), int(round(correct_pos[1]))
    tx, ty = int(round(target_pos[0])), int(round(target_pos[1]))
    
    if 0 <= wx < width and 0 <= wy < height:
        grid[wy][wx] = 'W'
    if 0 <= cx < width and 0 <= cy < height:
        if grid[cy][cx] == 'W':
            grid[cy][cx] = 'B'  # both at same cell
        else:
            grid[cy][cx] = 'C'
    if 0 <= tx < width and 0 <= ty < height:
        if grid[ty][tx] in ('W','C','B'):
            grid[ty][tx] = 'X'
        else:
            grid[ty][tx] = 'T'
    for row in grid:
        print(''.join(row))
    print("Legend: W=Wrong robot, C=Correct robot, T=Target, X=Robot+Target, B=Both robots")

def move_toward(robot_pos, target_pos, step_size=0.5):
    to_target = [target_pos[0]-robot_pos[0], target_pos[1]-robot_pos[1]]
    dist = norm(to_target)
    if dist < step_size:
        return target_pos[:]
    direction = unit_vector(to_target)
    return [robot_pos[0] + direction[0]*step_size,
            robot_pos[1] + direction[1]*step_size]
````

````python
# ---- Simulation parameters ----
robot_start = [2.0, 2.0]
robot_heading = 45.0
local_measurement = [4.0, 0.0]   # target is 4 units straight ahead

# Compute correct world target
correct_target = local_to_world(local_measurement, robot_start, robot_heading)
# Wrong robot uses local measurement as world target
wrong_target = local_measurement[:]

print(f"Robot start: {robot_start}, heading {robot_heading}°")
print(f"Local measurement: {local_measurement}")
print(f"Wrong robot thinks target = {wrong_target}")
print(f"Correct robot knows target = {correct_target}")

wrong_pos = robot_start[:]
correct_pos = robot_start[:]

for step in range(12):
    clear_output(wait=True)
    print(f"Step {step+1}")
    draw_map(wrong_pos, correct_pos, correct_target, width=12, height=8)
    
    wrong_pos = move_toward(wrong_pos, wrong_target, step_size=0.5)
    correct_pos = move_toward(correct_pos, correct_target, step_size=0.5)
    
    dist_correct = norm([correct_target[0]-correct_pos[0], correct_target[1]-correct_pos[1]])
    if dist_correct < 0.3:
        print("\n✅ Correct robot reached the target!")
        break
    time.sleep(0.4)
else:
    print("\nMax steps reached.")
print(f"\nFinal wrong robot: {wrong_pos}")
print(f"Final correct robot: {correct_pos}")
````

## 6. Visualising the robot’s local axes (extra)

To understand why the correct robot succeeds, let’s draw the robot’s local X (forward) and Y (left) axes on the map. This makes it clear that the local measurement `[4,0]` points along the local X axis, which after rotation becomes a specific world direction.

**Note:** This cell adds simple ASCII arrows to show the axes. It is optional but helpful.

````python
def draw_with_axes(robot_pos, heading_deg, width=12, height=8):
    """Draw robot position and its local X and Y axes as arrows."""
    grid = [['.' for _ in range(width)] for _ in range(height)]
    rx, ry = int(round(robot_pos[0])), int(round(robot_pos[1]))
    if 0 <= rx < width and 0 <= ry < height:
        grid[ry][rx] = 'R'
    # compute local X and Y axis endpoints in world
    R = rotation_matrix_2d(heading_deg)
    # local X (forward) = (1,0) rotated
    forward = matrix_vector_multiply(R, [1,0])
    # local Y (left) = (0,1) rotated
    left = matrix_vector_multiply(R, [0,1])
    # draw simple arrows: we just mark cells one step away
    fx, fy = int(round(rx + forward[0])), int(round(ry + forward[1]))
    lx, ly = int(round(rx + left[0])), int(round(ry + left[1]))
    if 0 <= fx < width and 0 <= fy < height:
        if grid[fy][fx] == '.': grid[fy][fx] = '↑'
    if 0 <= lx < width and 0 <= ly < height:
        if grid[ly][lx] == '.': grid[ly][lx] = '←'
    for row in grid:
        print(''.join(row))
    print("R = robot, ↑ = forward direction, ← = left direction")

# Example: robot at (5,4) heading 45°
draw_with_axes([5,4], 45, width=12, height=8)
````

## 7. Step‑by‑step explanation of `local_to_world` code

Let’s go through the function line by line:

```python
def local_to_world(local_point, robot_pos, robot_heading_deg):
    R = rotation_matrix_2d(robot_heading_deg)
```
- Builds a 2x2 rotation matrix from the angle (converts to radians inside).

```python
    rotated = matrix_vector_multiply(R, local_point)
```
- Multiplies the matrix by the local vector. This rotates the local vector so that it aligns with world axes.
- Example: if robot heading is 45°, local (4,0) → rotated ≈ (2.828, 2.828).

```python
    world_point = [rotated[0] + robot_pos[0], rotated[1] + robot_pos[1]]
```
- Adds the robot’s world position to the rotated vector – translation.
- Example: robot at (2,2) + rotated (2.828,2.828) = (4.828, 4.828) – the true target world position.

```python
    return world_point
```
- Returns the target’s coordinates in the world frame.

## 8. Why the wrong robot still moves but never reaches the correct target

The wrong robot **does** have a target – it’s just the wrong one. It will happily move toward its mistaken target (e.g., (4,0) if the local measurement was [4,0]).

If the robot’s starting position is not at the origin, and its heading is not zero, then the true target is somewhere else entirely. The wrong robot will wander off and may even move away from the true target.

In the animation you will see the wrong robot (W) moving in a seemingly random direction while the correct robot (C) goes straight to the target (T).

This is analogous to a robot with misconfigured sensor extrinsics: it receives correct sensor data but interprets it in the wrong reference frame, causing it to behave incorrectly.

## Summary

- **Rotation matrix** derived from basis vectors: columns are transformed X and Y axes.
- **Coordinate transformation** = rotate then translate.
- **Wrong robot** ignores rotation → fails.
- **Correct robot** applies transformation → succeeds.

This is the foundation of **sensor fusion** and **VIO**.
