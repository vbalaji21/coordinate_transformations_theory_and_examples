# Day 2: Rotation Matrices & Coordinate Transformations
## Why robots need to transform sensor data

A robot’s sensors (camera, LiDAR, IMU) measure the world **relative to the robot itself** (local frame).
To navigate, the robot must convert those measurements into a **world coordinate frame**.

In this notebook you will:
- Implement 2D rotation matrices.
- Learn the difference between **local** and **world** frames.
- Simulate two robots chasing a target:
  - **Wrong robot**: uses sensor measurements directly as world coordinates → fails.
  - **Correct robot**: rotates local measurements into world coordinates → succeeds.

This builds **intuition** for **sensor fusion, extrinsic calibration, and VIO**.

## 1. Recap of Day 1: Linear Algebra Basics and vector functions
We reuse the same pure‑Python implementations (no NumPy).

```python
import math
import time
from IPython.display import clear_output

# ----- Day 1 functions (without numpy) -----
def dot_product(v1, v2):
    return sum(a*b for a,b in zip(v1,v2))

def norm(v):
    return math.sqrt(sum(x*x for x in v))

def unit_vector(v):
    n = norm(v)
    if n == 0: raise ValueError("Zero vector")
    return [x/n for x in v]

def matrix_multiply(A, B):
    rows_A, cols_A = len(A), len(A[0])
    rows_B, cols_B = len(B), len(B[0])
    if cols_A != rows_B: raise ValueError("Incompatible")
    result = [[0 for _ in range(cols_B)] for _ in range(rows_A)]
    for i in range(rows_A):
        for j in range(cols_B):
            total = 0
            for k in range(cols_A):
                total += A[i][k] * B[k][j]
            result[i][j] = total
    return result

def matrix_vector_multiply(A, v):
    rows, cols = len(A), len(A[0])
    if cols != len(v): raise ValueError("Matrix columns != vector length")
    return [sum(A[i][j] * v[j] for j in range(cols)) for i in range(rows)]
```

## 2. Rotation Matrix in 2D

**Formula:**  
A rotation by angle θ (counter‑clockwise) is:

**Why counter-clockwise**
It's convention because the standard unit circle places angles starting from the positive x‑axis and increasing toward the positive y‑axis - that’s CCW. It’s consistent with right‑hand rule in 3D.

**What does a rotation do ?**
It does a **linear transformation** (write what is a linear transformation in few words) that maps the original x-axis (1,0) and (0,1) to new rotated vectors. Intuition - It basically changes the reference frame by rotating along the Z-axis on a plane. 

**Where does the x‑axis (1,0) go after rotation by angle θ?**
- The rotated x‑axis still has length 1
- Its coordinates are (cosθ, sinθ) - because on the unit circle, that’s exactly the point at angle θ from the x‑axis.

$$
R(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}
$$

**Intuition:**  
This matrix rotates any vector (or point) by θ around the origin.  
Columns are unit vectors (orthogonal).  
Multiplying a point `[x, y]` by `R(θ)` gives the rotated point.

**Example:**  
Rotate (1,0) by 90° (θ = π/2):  
R(90°) = [[0, -1], [1, 0]] → (0, 1). Correct.

```python
def rotation_matrix_2d(angle_deg):
    """
    Returns a 2x2 rotation matrix for a given angle in degrees.
    """
    theta = math.radians(angle_deg)
    cos_t = math.cos(theta)
    sin_t = math.sin(theta)
    return [[cos_t, -sin_t],
            [sin_t,  cos_t]]

# Test: rotate (1,0) by 90°
R90 = rotation_matrix_2d(90)
point = [1,0]
rotated = matrix_vector_multiply(R90, point)
print(f"Rotating (1,0) by 90° → {rotated}")  # should be approx (0,1)
```

## 3. Transforming a point from local to world frame

Assume:
- The robot is at `(rx, ry)` in world coordinates.
- Its heading angle is `θ` (0 = facing east, increasing counter‑clockwise).
- A sensor measures the target **relative to the robot** as `(dx_local, dy_local)`.

**To get the target’s world coordinates:**
1. Rotate the local vector by θ (so it aligns with world axes).
2. Add the robot’s world position.

Formula:  
`target_world = R(θ) · [dx_local, dy_local] + [rx, ry]`

**Why the rotation?**  
Because the sensor’s measurement is expressed in the robot’s own coordinate frame (e.g., "target is 2 meters in front of me").  
If the robot is rotated, “front” points in a different world direction.

```python
def local_to_world(local_point, robot_pos, robot_heading_deg):
    """
    Convert a point given in robot-local coordinates to world coordinates.
    - local_point: [dx, dy] in robot frame
    - robot_pos: [rx, ry] world position of robot
    - robot_heading_deg: robot's orientation (0 = east, CCW)
    """
    R = rotation_matrix_2d(robot_heading_deg)
    rotated = matrix_vector_multiply(R, local_point)   # step 1
    world_point = [rotated[0] + robot_pos[0], rotated[1] + robot_pos[1]]  # step 2
    return world_point
```

## 4. The wrong vs correct robot simulation

**Scenario:**  
- Two robots start at the same world position `(2,2)` with heading 45°.
- A sensor gives the **local** target vector (e.g., `[3, 0]` = 3 meters straight ahead).
- The **true** target world position is computed correctly with the above function.

**Wrong robot:**  
- Ignores its own orientation → uses `local_target` directly as world target.
- Moves toward that incorrect point.

**Correct robot:**  
- Applies rotation + translation → finds correct world target.
- Moves correctly.

**Visualisation:**  
We will simulate both robots step by step and display an ASCII map.

```python
# Helper: draw robots, target on a grid
def draw_map(wrong_pos, correct_pos, target_pos, width=10, height=6):
    grid = [['.' for _ in range(width)] for _ in range(height)]
    # place robots
    wx, wy = int(round(wrong_pos[0])), int(round(wrong_pos[1]))
    cx, cy = int(round(correct_pos[0])), int(round(correct_pos[1]))
    tx, ty = int(round(target_pos[0])), int(round(target_pos[1]))
    
    if 0 <= wx < width and 0 <= wy < height:
        grid[wy][wx] = 'W'   # wrong robot
    if 0 <= cx < width and 0 <= cy < height:
        if grid[cy][cx] == 'W':
            grid[cy][cx] = 'B'  # both at same cell
        else:
            grid[cy][cx] = 'C'   # correct robot
    if 0 <= tx < width and 0 <= ty < height:
        if grid[ty][tx] in ('W','C','B'):
            grid[ty][tx] = 'X'   # target with robot
        else:
            grid[ty][tx] = 'T'
    # print
    for row in grid:
        print(''.join(row))
    print("Legend: W=Wrong robot, C=Correct robot, T=Target, X=Robot+Target, B=Both robots")
```

## 5. Simulation: both robots move toward the same target

We give both robots the **same local sensor measurement** (e.g., target is 4 units ahead in the robot’s local frame).
- **Wrong robot** treats that as world coordinates → `target_world_wrong = local_measurement`
- **Correct robot** applies rotation → `target_world_correct = R·local + robot_pos`

Then each robot moves one step toward **its** target estimate (using unit vector and move).

```python
def move_toward(robot_pos, target_pos, step_size=1.0):
    """Move robot one step toward target. Returns new position."""
    to_target = [target_pos[0]-robot_pos[0], target_pos[1]-robot_pos[1]]
    dist = norm(to_target)
    if dist < step_size:
        return target_pos[:]
    direction = unit_vector(to_target)
    return [robot_pos[0] + direction[0]*step_size, robot_pos[1] + direction[1]*step_size]
```

```python
# ---- Simulation parameters ----
robot_start = [2.0, 2.0]
robot_heading = 45.0          # degrees, 0 = east
local_target_measurement = [4.0, 0.0]   # sensor says: target is 4 units straight ahead

# Compute correct world target using transformation
correct_target_world = local_to_world(local_target_measurement, robot_start, robot_heading)

# Wrong robot uses local measurement AS world coordinates
wrong_target_world = local_target_measurement[:]   # [4,0]

print(f"Robot start: {robot_start}, heading: {robot_heading}°")
print(f"Local measurement: {local_target_measurement}")
print(f"Wrong robot thinks target is at: {wrong_target_world}")
print(f"Correct robot knows target is at: {correct_target_world}")

# Initialize positions
wrong_pos = robot_start[:]
correct_pos = robot_start[:]

# ---- Animated simulation ----
for step in range(12):
    clear_output(wait=True)
    print(f"Step {step+1}")
    draw_map(wrong_pos, correct_pos, correct_target_world, width=12, height=8)
    
    # Move wrong robot toward its (incorrect) target
    wrong_pos = move_toward(wrong_pos, wrong_target_world, step_size=0.5)
    # Move correct robot toward the true world target
    correct_pos = move_toward(correct_pos, correct_target_world, step_size=0.5)
    
    # Check if correct robot reached target
    dist_correct = norm([correct_target_world[0]-correct_pos[0], correct_target_world[1]-correct_pos[1]])
    if dist_correct < 0.3:
        print("\n✅ Correct robot reached the target!")
        break
    time.sleep(0.4)
else:
    print("\nMax steps reached.")
    
print(f"\nFinal wrong robot position: {wrong_pos}")
print(f"Final correct robot position: {correct_pos}")
```

## 6. Why the wrong robot fails – explanation

The wrong robot **ignores its own orientation** and treats every sensor measurement as if the robot were always facing east (0°).

When the robot is rotated, the sensor’s “forward” points in a different world direction.  
Without applying the rotation, the wrong robot will chase a ghost target – it will never reach the real target unless it randomly stumbles into it.

**This is exactly what happens in real robotics when:**
- Camera extrinsics are wrong (the camera’s pose relative to the robot is misconfigured).
- The IMU frame is not correctly aligned with the robot’s body frame.
- The LiDAR’s transform (TF) is missing or incorrect.

By contrast, the **correct robot** uses the rotation matrix to convert local measurements into the world frame, and therefore moves accurately toward the true target.
